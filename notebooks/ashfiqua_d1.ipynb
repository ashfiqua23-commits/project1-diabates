# data cleaning

count missing calues per column

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

express missingness as a proportion

```python
df.isnull().mean()
```

inspect the affecte rows

```python
df[df['physical_activity'].isnull()]
df[df.isnull().any(axis=1)]
```

count the placeholder

```python
(df['glucose']==0).sum()
```

```python
placeholder_cols=["glucose","blood_pressure","skin_thickness","insulin","bmi"]
for c in placeholder_cols:
    print(c,(df[c]==0).sum(),"zero")
```

convert placeholders to missing value

```python
import numpy as np
```

```python
df["glucose"]=df["glucose"].replace(0,np.nan)
```

```python
for c in placeholder_cols:
    df[c]=df[c].replace(0,np.nan)
```

```python
df.isnull().sum().sum()
```

```python
df.isnull().sum()
```

```python
df.isnull().mean()
```

fill numeric columns

```python
df["glucose"]=df["glucose"].fillna(df["glucose"].median())
df["glucose"]=df["glucose"].fillna(df["glucose"].mean())
```

fill numerical columns

```python
col_fill=["age","blood_pressure","skin_thickness","insulin","bmi"]
for c in col_fill:
    df[c]=df[c].fillna(df[c].median())
```

```python
df.isnull().sum()
```

fill categorical columns

```python
fill_columns=["smoker","physical_activity"]
for c in fill_columns:
    df[c]=df[c].fillna("unknown")
```

verify no missing values remain

```python
df.isnull().sum().sum()
```

```python
df["gender"].unique()
```

remove stray whitespace

```python
df["gender"]=df["gender"].str.strip()
```

```python
df.value_counts("gender")
```

```python
smkr_list=["smoker","physical_activity"]
for c in smkr_list:
    df[c]=df[c].str.strip()
```

standardise capitalization


```python
df["gender"]=df["gender"].str.lower()
```

```python
df.value_counts("gender")
```

```python
df["physical_activity"]=df["physical_activity"].str.title()
```

```python
df.value_counts("physical_activity")
```

```python
df["smoker"]=df["smoker"].str.upper()
```

unify equivalent categories

```python
df["gender"]=df["gender"].replace({'f':'female','m':'male'})
```

```python
df.value_counts("gender")
```

```python
df["smoker"]=df["smoker"].replace({'Y':'YES','N':'NO'})      
```

```python
df.value_counts("smoker")
```

convert text dates to real dates

```python
df["last_checkup_date"]=pd.to_datetime(df["last_checkup_date"])
df["last_checkup_date"]=pd.to_datetime(df["last_checkup_date"],dayfirst=True)
```

correct remaining type mismathes

```python
df["age"]=df["age"].round().astype(int)
df.value_counts('age')
```
